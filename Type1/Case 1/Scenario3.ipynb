When the function does not check the bounds of that array, it can access memory before or
after the buffer.

### Pattern Info

Variables:
- FormSpec = The format specifier for the given datatype
- Datatype = Default datatypes in C
- GeneratedNumber = Random generated number
- GeneratedNumber2 = Random generated number


These constraints must be met:
- GeneratedNumber > 0
- GeneratedNumber2 >= GeneratedNumber or GeneratedNumber2 < 0
- Datatypes can be any of the following:
    - int, float, double, short, long, long long, unsigned
- The format specifier must match the given datatype

```python
from pathlib import Path
import random


folder = Path('..') / '..' / 'GeneratedCode'

datatypes = ["int", "float", "double", "short", "long", "long long", "unsigned int", "unsigned short", "unsigned long", "unsigned long long"]

FormSpecDict = {"int": "d", "float": "f", "double": "f", "short": "hd", "long": "ld", "long long": "lld", "unsigned int": "u", "unsigned short": "hu", "unsigned long": "lu", "unsigned long long": "llu"}

filesGenerated = 10
for i in range(filesGenerated):
    fileName = 'type1_case1_scenario3_' + str(i) + '.c'
    file = folder / fileName
    generatedNumber = random.randint(3, 5000)
    randomType = random.randint(0, 1)
    generatedNumber2 = 0
    if (randomType == 0):
        generatedNumber2 = random.randint(generatedNumber, 5000)
    else:
        generatedNumber2 = -1 * random.randint(0, 5000)
    dataType = datatypes[random.randint(0, len(datatypes) - 1)]
    formSpec = FormSpecDict[dataType]
    my_c_code = '#include <stdio.h>\n\nvoid printValue(' + dataType +' * array, index) {\n    printf(\"Value: %' + formSpec + '\", array[index]);\n}\n\nint main () {\n    ' + dataType + ' myArray[' + str(generatedNumber) + '];\n    int i = 0;\n    for (i = 0; i < ' + str(generatedNumber) + '; i++) {\n        myArray[i] = 1;\n    }\n    printValue(myArray, ' + str(generatedNumber2) + ');\n    return 0;\n}\n\n' 
    file.write_text(my_c_code)
```
